```python
import pandas as  pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel("tg.xlsx")

df = df['Текст публикации']
```

```python
posts = list(df)
```

```python
results = model.predict(posts, k=1)
```

```python
all_sentiments = []

for r in results:
  sentiment = list(r.keys())
  if ('skip' not in sentiment and 'speech' not in sentiment):
    all_sentiments.append(sentiment[0])
```

```python
from collections import Counter

Counter(all_sentiments)
```

```python
plt.hist(all_sentiments)

plt.show()
```
