```python
from natasha import (
    Segmenter,

    NewsEmbedding,
    NewsMorphTagger,
    NewsSyntaxParser,

    Doc
)
segmenter = Segmenter()

emb = NewsEmbedding()
morph_tagger = NewsMorphTagger(emb)
syntax_parser = NewsSyntaxParser(emb)
```

```python
# телеграм-пост из имеющегося задела данных
text = 'ЧВК РЕДАН — НЕ СУЩЕСТВУЕТ. Нет организации реданов, нет такой субкультуры. Услышала это вчера, ничего не понятно, но самое главное — это очень тупо. Идиотизм, как по мне. Это диз информация, с помощью которой люди сеют хаос вокруг себя. Сами себе же создаёте проблемы, сами же боитесь. Нельзя бить людей за их внешний вид: хочется им выглядеть так - это их дело, я и до этого твердила это очень долго. Если вам так хочется участвовать во всяких кланах '

doc = Doc(text)
```

```python
doc.segment(segmenter)
doc.tag_morph(morph_tagger)
doc.parse_syntax(syntax_parser)
```

```python
doc.sents[0]
```

```python
doc.sents[0].syntax.print()
```

```python
from natasha import NewsNERTagger

ner_tagger = NewsNERTagger(emb)
doc.tag_ner(ner_tagger)
doc.ner.print()
```

```python
from natasha import MorphVocab

morph_vocab = MorphVocab()

for token in doc.tokens:
  token.lemmatize(morph_vocab)

{_.text: _.lemma for _ in doc.tokens}
```
