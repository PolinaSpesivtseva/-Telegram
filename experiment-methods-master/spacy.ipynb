```python
from spacytextblob.spacytextblob import SpacyTextBlob
nlp.add_pipe('spacytextblob')
```

```python
nlp.pipe_names
```

```python
# телеграм-пост из канала Дмитрия Медведева

text = 'Деятель в грязной зелёной майке сказал, что россияне серьёзно почувствуют результаты их работы. Он бы, урод с напудренным носом, лучше съездил и посмотрел на результаты работы российских Вооружённых сил по целям в Одессе, Днепропетровске, Львове, Харькове, Измаиле, Новомихайловке, Ильичёвске и др. после террористических атак Украины. И задумался, что он сделал со своей страной!'
```

```python
doc = nlp(text)
```

```python
for token in doc:
  print(token.text, token.pos_)
```

```python
for token in doc:
    print(token.text, token.lemma_)
```

```python
for ent in doc.ents:
    print(ent.text, ent.label_)
```

```python
doc._.blob.polarity
```

```python
doc._.blob.subjectivity
```

```python
doc._.blob.sentiment_assessments.assessments
```

```python
from textblob import TextBlob

analysis = TextBlob(text)
```

```python
sentiment = analysis.sentiment
```

```python
print(sentiment.polarity)
```
